# 🦓 Project ZEBRA: Exploratory Data Analysis

Welcome to **ZEBRA** (Zero-Error Bias Risk Assessment). This notebook serves as an interactive introduction to the Health Insurance Claim Prediction dataset. We will explore the distribution of our target variable (whether a claim was filed), inspect the characteristics of our numerical features, and discover the underlying correlations.

Let's dive into the data story.

```python
import sys
from pathlib import Path
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# Add the parent directory to Python path to import our modules
sys.path.append(str(Path.cwd().parent))

from src.config import load_config
from src.visualization import (
    plot_target_distribution,
    plot_numeric_distributions,
    plot_correlation_heatmap
)
```

## 1. Load Configuration and Data
We use a centralized `settings.yaml` configuration file to manage our data paths and feature lists cleanly.

```python
# Load settings
config = load_config('../config/settings.yaml')
train_path = Path('../') / config['data']['train_path']

# Load the dataset
print(f"Loading data from: {train_path}...")
df = pd.read_csv(train_path)
print(f"Dataset Shape: {df.shape[0]:,} rows and {df.shape[1]} columns.")
```

## 2. Target Variable Distribution
The target variable indicates whether a health insurance claim was filed (`1`) or not (`0`). Let's visualize the balance of our classes.

```python
target_col = config['features']['target']
plot_target_distribution(df, target_col)
```

> **Insight**: As expected with insurance datasets, there is a severe class imbalance. We will handle this in our preprocessing pipeline using techniques like SMOTE and Undersampling.

## 3. Numeric Feature Distributions
Let's look at the distribution of the first few numeric features. Are they skewed? Are there heavy tails?

```python
numeric_cols = config['features']['numeric']
plot_numeric_distributions(df, numeric_cols)
```

## 4. Feature Correlations
Which features are highly correlated with one another? Highly correlated features might represent redundant information or provide candidates for interaction features.

```python
# Take the top 15 numeric features plus the target to keep the heatmap readable
heatmap_features = numeric_cols[:15] + [target_col]
plot_correlation_heatmap(df, heatmap_features)
```
